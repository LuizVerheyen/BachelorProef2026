```python
## Data minipulation
import pandas as pd

## Scraping
from bs4 import BeautifulSoup
import requests

import time
from datetime import datetime
```

```python
start_date = '2022-02-01'
end_date = datetime.now().strftime("%Y-%m-%d")
amount_of_tweets_per_page = 500
page_number = 1
```

```python
url = f'https://trumpstruth.org/search?query=&start_date={start_date}&end_date={end_date}&sort=date_desc&removed=include&per_page={amount_of_tweets_per_page}&page={page_number}'
```

```python
response = requests.get(url=url)
print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
print(soup.prettify()) # prints well formatted HTML
```

```python
# Zoek alle dates
full_date = soup.find_all('a', class_='status-info__meta-item')
```

```python
# Index 0 is de gebruikersnaam, Index 1 is de datum
if len(full_date) > 1:
    full_date = full_date[1].text.strip()
    print(full_date)# "April 8, 2026, 11:46 PM"
    
    # Splitsen op de laatste komma om datum en tijd te scheiden
    # We splitsen op ", " en pakken de delen
    parts = full_date.split(', ')
    print(parts)

    date = f"{parts[0]}"
    year = f"{parts[1]}"
    time = f"{parts[2]}"
    
    print(f"Date: {date}")
    print(f"year: {year}")
    print(f"time: {time}")
```

```python
deleted_badges = soup.find_all("div", class_='status__deleted-badge-wrap')
print(deleted_badges)
```

```python
if len(deleted_badges) > 1:
    print("found a deleted badge!")
```

```python
content = soup.find_all("div", class_='snippet-clean-content')
```

```python
if len(content) > 1:
    print(content[0].text.strip())
```

```python
post_div = soup.find('div', class_='search-result')
```

```python
if post_div:
    url = post_div.get('data-status-url')
    print(url)
```
