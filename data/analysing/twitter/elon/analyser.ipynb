```python
import pandas as pd
import sys
from pathlib import Path
```

```python
ROOT = Path().resolve().parents[3]
sys.path.append(str(ROOT))

print(ROOT)
```

```python
DATASET = ROOT / 'data' / 'raw' / 'tweets' / 'elonmusk' / 'all_musk_posts2010-2025.csv'
```

```python
df = pd.read_csv(DATASET)
df.head(10)
```

```python
df.info()
```

```python
df.describe()
```

```python
for col in df.columns:
    missing_pct = df[col].isna().mean() * 100
    print(f"{col}: {missing_pct:.2f}% missing")
```

```python
df = df[['createdAt','url','fullText','retweetCount','replyCount','likeCount']]
```

```python
df.info()
```

```python
df.rename(columns={
    'retweetCount' : "Reposts",
    'replyCount' : 'Comments',
    'likeCount' : 'Likes'
},inplace=True)
```

```python
df = df[~df['fullText'].str.startswith("RT", na=False)]
```

```python
for i in df['url']:
    print(i)
```

```python
for col in df.columns:
    missing_pct = df[col].isna().mean() * 100
    print(f"{col}: {missing_pct:.2f}% missing")
```

```python
df_still_need_searching = df[
    (df['Likes'] == 0) &
    (df['Comments'] == 0) &
    (df['Reposts'] == 0)
]
```

```python
df.size
```

We moeten nog 366 tweets scrapen om de metadate hier van op te halen &rarr; het kan zijn dat deze gewoon al ge delete zijn

```python
df_still_need_searching.size
```

We kunnen zien na deze check dat de url's zonder metadata tweets zijn NIET van elonmusk dus kunnen we ze gewoon verwijderen

```python
for url in df_still_need_searching['url']:
    if "elonmusk" in str(url).lower():
        print(url)
```

```python
df = df[df['url'].str.contains("elonmusk", case=False, na=False)]
```

```python
df.size
```

```python
df['createdAt'] = pd.to_datetime(df['createdAt'])

df['DateKey'] = df['createdAt'].dt.strftime('%Y%m%d').astype(int)
df['TimeKey'] = df['createdAt'].dt.strftime('%H%M%S').astype(int)
```

```python
df.drop(columns=['createdAt'],inplace=True)
```

```python
df.head(2)
```

```python
for col in df.columns:
    missing_pct = df[col].isna().mean() * 100
    print(f"{col}: {missing_pct:.2f}% missing")
```

```python
df = df[
    ~(
        (df['Likes'].isna()) &
        (df['Comments'].isna()) &
        (df['Reposts'].isna())
    )
]
```

```python
df.shape
```

```python
df.rename(columns={
    'fullText' : 'Text'
},inplace=True)
```

```python
df.head()
```

```python
df.to_csv(ROOT / "data" / "processed" / "elonmusk.csv", index=False)
```
