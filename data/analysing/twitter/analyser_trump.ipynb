```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../raw/trump_tweets.csv")
```

```python
df.head()
```

```python
for i in df.columns:
    if df[i].dtype == "float64":
        resultaat = df[df[i] > 0][i].count()
    else:
        resultaat = df[df[i] != None][i].count()
        
    print(f"{i}: {resultaat / len(df) * 100}% filled")
    
```

```python
df.drop(columns=["id","platform", "word_count", "urls", "post_url", "handle", "media_urls", "hashtags"], inplace=True)
```
