## Filtering data

#### we gaan data eruit halen waarbij de text empty is 
#### we houden ook enkel de date, text, username en url

```python
df = df[['date', 'handle','text', 'post_url']].dropna(subset=['text'])
```

```python
print(f"total tweets after filtering of twitter: {len(df)}")
```

##### all text columns has a value

```python
print(len(df[df['text'] == '']))
```

we also want to delete all the tweets that only have a link as the text

```python
df.head()
```

```python
import re

df = df[
    ~df['text'].str.fullmatch(r'https?://\S+', na=False)
]
```

```python
print(f"total tweets after filtering of twitter: {len(df)}")
```

```python
df.head()
```

```python
df.columns = [col.capitalize() for col in df.columns]
```

### make datekey,TimeKey and renaming columns to match database

```python
df['Date'] = pd.to_datetime(df['Date'])


df['DateKey'] = df['Date'].dt.strftime('%Y%m%d')
df['TimeKey'] = df['Date'].dt.strftime('%H%M%S')
```

```python
df.drop(columns=['Date'],inplace=True)
```

```python
df.rename(columns={
    'Post_url' : 'url',
    'Handle' : "UserName"
},inplace=True)
```

```python
df.head()
```

```python
df['DateKey'] = df['DateKey'].astype(int)
df['TimeKey'] = df['TimeKey'].astype(int)
```

### save

```python
df.to_csv("../../processed/trump_twitter.csv", index=False)
```
